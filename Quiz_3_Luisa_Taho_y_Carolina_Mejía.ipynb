```python
import numpy as np
import matplotlib.pyplot as plt

# El siguiente código se construyó basándonos en el código del quiz anterior, documentación y los vídeos de clase.

# Se definen las condiciones iniciales, constantes
P = 1
g = 1

ñ = 325280

# Definición de vectores de valores
r_1 = np.linspace(1, 15, 170)  # Vector de distancias radiales
t_1 = np.linspace(0, 400, 100)  # Vector de ángulos en grados

# Como queremos graficar el potencial eléctrico, definimos la función que contiene nuestra solución analítica
def Pot_Electrico(P, t, g, r):
    # Función que calcula el potencial eléctrico dado P, t, g, y r
    return np.cos(np.radians(t)) * P / (4 * g * np.pi * r**2)

# Generación de valores para los ejes del gráfico
valores_radio = np.linspace(1, 5, 100)  # Valores radiales de 1 a 5, en 100 puntos
valores_t = np.linspace(0, 360, 100)  # Valores angulares de 0 a 360 grados, en 100 puntos

# Creación de una malla de valores de r y t para evaluar el potencial en una rejilla
comp_t, comp_r = np.meshgrid(valores_t, valores_radio)
# Cálculo del potencial eléctrico en cada punto de la malla
Potencial = Pot_Electrico(P, comp_t, g, comp_r)

# Gráfico de líneas equipotenciales
plt.figure(figsize=(20, 35))  # Tamaño de la figura
plt.contour(comp_r * np.cos(np.radians(comp_t)),
            comp_r * np.sin(np.radians(comp_t)),
            Potencial, levels=50)  # Líneas equipotenciales con 50 niveles

# Configuración del gráfico
plt.title('Distribución del Potencial Eléctrico (φ)', size=20)  # Título del gráfico
plt.xlabel('Eje R', size=15)  # Etiqueta del eje X
plt.ylabel('Eje X', size=15)  # Etiqueta del eje Y
plt.colorbar(label='Potencial eléctrico (φ)')  # Barra de color que indica los valores del potencial

# Mostrar el gráfico
plt.show()


```
